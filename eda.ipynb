```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
```

```python
df = pd.read_csv('Spotify_Dataset_V3 2.csv', delimiter=';')
```

```python
df.head()

```

1. Drop unused column EX. Song URL

```python
df.drop(['Song URL'],axis = 1, inplace = True)
```

2. Check Data Profile

```python
df[['Danceability', 'Energy', 'Loudness', 'Speechiness', 'Acousticness', 'Instrumentalness', 'Valence']].describe()
```

3. We need to normalize data for loudness first

```python
cleaned_df = df[(df['Loudness'] <= 0 ) & (df['Loudness'] >= -60 )] # This is 
cleaned_df
```

```python
df['Loudness'] = np.where(
    (df['Loudness'] <= -60) | (df['Loudness'] >= 0),  # Condition: x is NOT in (-60, 0)
    df['Loudness'] / 1000,                           # Value if TRUE (divide the "others")
    df['Loudness']                                   # Value if FALSE (keep the "good" values)
)
```

```python
df['Loudness'] = MinMaxScaler().fit_transform(df[['Loudness']])
```

4. Change DateTime

```python
df['Date'] = pd.to_datetime(df['Date'], format='%d/%m/%Y')
```

```python
df
```

```python
is_duplicate = df.duplicated(subset=['Rank', 'Date', 'Title'], keep='first')
duplicates = df[is_duplicate]
df.drop(duplicates.index, inplace=True)
```
